In this chapter we will see how the mathematics behind the spin in quantum mechanics is deeply connected to the Lie algebra sl(2) that we have been studying. The idea is not to derive the physics of spin from scratch. Instead, we will take a few basic facts about spin as given, and watch how they force us — step by step — into exactly the machinery we already built.

## spin

Let's assume a spin-1/2 particle, such as an electron. The spin of the electron is a quantum property that can be measured along any axis in physical space. When we measure the spin along the vertical axis, we can only get one of two possible outcomes: "spin up" or "spin down". These outcomes can be represented by column vectors:

$$
v_\text{up} = \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \quad v_\text{down} = \begin{pmatrix} 0 \\ 1 \end{pmatrix}.
$$

Before we measure a particle's spin, we can describe its state as a linear combination (a superposition) of the two possible outcomes:

$$
v = \alpha v_\text{up} + \beta v_\text{down}.
$$

Here $\alpha$ and $\beta$ are **complex** numbers, called amplitudes. The rule connecting them to measurement (the Born rule) is that the probability of measuring "up" is $|\alpha|^2$ and the probability of measuring "down" is $|\beta|^2$. Since the particle must come out as one or the other, these probabilities have to add up to one:

$$
|\alpha|^2 + |\beta|^2 = 1.
$$

So a spin state is nothing more than a two-component complex vector of length one. Two remarks before we move on. First, the space is $2$-dimensional because we are describing a two-state system — that "2" is a fact about the electron's spin, not about any algebra. Keep an eye on it; it is *not* the same "2" that will show up in the size of our matrices, even though they are about to coincide. Second, "length one" here means length in the complex sense: $\lVert v \rVert^2 = |\alpha|^2 + |\beta|^2$, which is exactly $v^\dagger v$, where $v^\dagger$ is the conjugate transpose (the [bra](/quantum/bra_ket.html) $\langle v|$). The complex conjugation is what makes this a real, non-negative number that we are entitled to call a total probability.

## transforming spin states

Now let's ask the question that will drive the rest of the chapter: **what can we do to a spin state?** We can let it evolve in time, or rotate the measuring apparatus, or apply a magnetic field. Whatever the physical operation is, it acts on the state as some linear operator $M$:

$$
v \;\longmapsto\; M v.
$$

But not every $M$ is allowed. Whatever the state turns into, it is still a state of the same particle, so it must *still* describe a valid set of probabilities — its total probability must still be one. In other words, $M$ has to preserve the length of every state:

$$
\lVert M v \rVert = \lVert v \rVert \quad \text{for every } v.
$$

This one physical requirement pins down what kind of operator $M$ can be. Let's write the length-preservation condition out using $\lVert w \rVert^2 = w^\dagger w$:

$$
(Mv)^\dagger (Mv) = v^\dagger v
\quad\Longrightarrow\quad
v^\dagger \big(M^\dagger M\big) v = v^\dagger v.
$$

If this holds for *every* vector $v$, the operator sandwiched in the middle must be the identity:

$$
M^\dagger M = I.
$$

An operator satisfying $M^\dagger M = I$ is called **unitary** — its conjugate transpose is its inverse. So the conclusion is clean and forced: *the transformations of a spin state are exactly the unitary operators.* The physical demand "conserve total probability" and the mathematical condition "unitary" are the very same statement, one wearing physics clothes and the other wearing linear-algebra clothes.

The set of all $2\times 2$ unitary matrices forms a group, written $U(2)$. We are almost home — but not quite.

## fixing the phase: from $U(2)$ to $SU(2)$

There is one more physical fact to fold in, and it will cut $U(2)$ down to the group we actually want.

A global phase is invisible. If we multiply the entire state by a phase factor $e^{i\varphi}$, sending $v \mapsto e^{i\varphi} v$, then every probability $|\alpha|^2, |\beta|^2$ is unchanged (the phase has modulus one), and so is every relative phase between components. No measurement can tell $v$ from $e^{i\varphi} v$. Physically, they are the same state.

This freedom lets us tidy up our transformations. A unitary matrix always has a determinant that is a pure phase — from $M^\dagger M = I$ we get $|\det M|^2 = 1$, so $\det M = e^{i\varphi}$ for some real $\varphi$. Because an overall phase is physically meaningless, we are free to multiply $M$ by a phase $\lambda$ (with $|\lambda| = 1$) chosen so that the determinant becomes exactly one: we just need $\lambda^2 \det M = 1$, which we can always solve. The rescaled matrix is still unitary, and now it also has determinant one.

So, losing no physics, we may always represent a transformation by a matrix that is **unitary and has determinant one**. This is the special unitary group $SU(2)$:

$$
SU(2) = \left\{\, U \;\middle|\; U^\dagger U = I,\ \det U = 1 \,\right\}.
$$

Notice the order of the argument, because it matters. Unitarity is the real physical constraint, coming from conservation of probability. Determinant one is a *convenience*, a way of using the invisible-phase freedom to pick a tidy representative. We did not impose it as a second physical law.

(One honest caveat, which will grow into the whole point of the last section. Fixing the determinant to one does not use up *all* of the phase freedom: the phase $\lambda$ that does the job is determined only up to a sign, so $U$ and $-U$ both survive with determinant one. That leftover sign is not an accident, and it is where spinors get strange. Hold that thought.)

## back to familiar ground: $su(2)$ as a tangent space

Here is the payoff for all the machinery we built earlier. We have arrived at $SU(2)$, a group defined by two conditions — determinant one and unitary — and we already know how to handle "determinant one," and we know exactly how to trade a constraint on a group for a constraint on its algebra: differentiate at the identity.

Write a transformation close to the identity as $U = e^{tX}$, where $X$ lives in the tangent space (the Lie algebra). We ask what each of our two conditions demands of $X$.

**Determinant one.** Using the identity $\det(e^{tX}) = e^{t\,\operatorname{tr} X}$ that we proved [earlier](/lie_algebra/basic_properties.html#detexp), the condition $\det U = 1$ for all $t$ becomes

$$
e^{t\,\operatorname{tr} X} = 1 \quad\Longrightarrow\quad \operatorname{tr} X = 0.
$$

This is the very same computation that gave us traceless matrices for $\mathfrak{sl}(2)$. Nothing new.

**Unitary.** Differentiate $U^\dagger U = I$ with respect to $t$ and evaluate at $t = 0$. Using $\frac{d}{dt}U = X e^{tX}$ (so $\dot U|_0 = X$) and $\frac{d}{dt}U^\dagger = X^\dagger e^{tX^\dagger}$ (so $\dot U^\dagger|_0 = X^\dagger$),

$$
\frac{d}{dt}\big(U^\dagger U\big)_{t=0} = \dot U^\dagger|_0\, \underbrace{U|_0}_{I} + \underbrace{U^\dagger|_0}_{I}\, \dot U|_0 = X^\dagger + X = 0.
$$

So the condition is

$$
X^\dagger = -X,
$$

which says $X$ is **anti-Hermitian**.

Put the two together: the Lie algebra of $SU(2)$ consists of matrices that are **traceless and anti-Hermitian**. This is $\mathfrak{su}(2)$, and — as we discussed when we studied real forms — it is exactly the [compact real form](/lie_algebra/real_forms.html#su2) of the $\mathfrak{sl}(2,\mathbb{C})$ we have been developing all along. We did not build a new object. Physics reached into $\mathfrak{sl}(2,\mathbb{C})$ and selected this particular real slice, the same way unitarity selected it before.

And notice the machine fired twice, identically: a multiplicative constraint on the group ("product equals identity") linearized into an additive constraint on the algebra ("sum equals zero"). Determinant one became trace zero; unitary became anti-Hermitian. Same trick, same page in the playbook.

Because $\mathfrak{su}(2)$ is a real form of $\mathfrak{sl}(2,\mathbb{C})$, *everything* we developed transfers: the bracket, the structure constants, the ladder operators, and above all the classification of representations. The two-component state $v$ we started this chapter with is just the standard (2-dimensional) representation. The other representations are the other spins — and by the dimension count we already derived, the representation of dimension $2j+1$ is spin $j$. Spin-$\tfrac12$ is the $2$-dimensional one, spin-$1$ the $3$-dimensional one, spin-$2$ the $5$-dimensional one, and so on. The "2" from the electron was only ever the smallest room in a building we had already mapped.

## the generators of $su(2)$

Let's write down a concrete basis, because we will need it in a moment. The traditional choice starts from the **Pauli matrices**:

$$
\sigma_1 = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}, \qquad
\sigma_2 = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}, \qquad
\sigma_3 = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}.
$$

These are Hermitian and traceless, so they are *not* themselves in $\mathfrak{su}(2)$ (whose elements are anti-Hermitian). But multiply each by $-i/2$ and you land inside the algebra. Define

$$
T_j = -\tfrac{i}{2}\,\sigma_j,
$$

which are traceless and anti-Hermitian, as required. Their brackets reproduce the structure constants we have seen before, using $[\sigma_i, \sigma_j] = 2i\,\epsilon_{ijk}\sigma_k$:

$$
[T_i, T_j] = \left(-\tfrac{i}{2}\right)^2 [\sigma_i, \sigma_j] = -\tfrac14\big(2i\,\epsilon_{ijk}\sigma_k\big) = \epsilon_{ijk}\, T_k.
$$

If you want to see how these sit inside the $E, H, F$ world we built earlier, just multiply out: $iH$, $E - F$, and $i(E+F)$ are precisely $i\sigma_3$, $i\sigma_2$, and $i\sigma_1$ — the same three combinations we identified as the real form when we first met $\mathfrak{su}(2)$. Nothing here is new; it is our old basis in physics notation.

One connection worth making explicit, since it closes a loop from the previous chapters. Physicists rarely carry the anti-Hermitian $T_j$ around. They prefer the **Hermitian** operators $S_j = \tfrac12 \sigma_j$, because those are *observables* — their eigenvalues $\pm\tfrac12$ are the actual measured values of spin along an axis, and observables must be Hermitian to have real eigenvalues. The algebra generator and the observable differ by exactly the factor we kept running into: $T_j = -i\, S_j$. The anti-Hermitian object generates the transformation; the Hermitian object is the thing you measure; the $i$ is the bridge between them.

## spinors

We now have everything we need to explain the object that gives this chapter its name.

A **spinor** is a two-component complex vector — a state like our $v$ — that transforms under $SU(2)$ in the standard representation. That definition sounds harmless. What makes spinors genuinely different from ordinary vectors, or from a classical arrow representing angular momentum, is how they respond to rotations of physical space.

It turns out that a rotation of physical space by an angle $\theta$ about a unit axis $\hat{n}$ is represented on spinors by the $SU(2)$ matrix

$$
U(\theta, \hat{n}) = \exp\!\left(-\,i\,\frac{\theta}{2}\,\hat{n}\cdot\vec\sigma\right).
$$

We will *derive* this correspondence — where it comes from, and why there is a factor of $\tfrac12$ sitting in the exponent — in the next chapter, when we work out the precise relationship between $SU(2)$ and the rotation group $SO(3)$. For now, take it as given and let's see what it does, because the consequence is immediate and striking.

Using $(\hat{n}\cdot\vec\sigma)^2 = I$, the exponential collapses to a clean closed form (the same way $e^{i\alpha}$ splits into sine and cosine):

$$
U(\theta, \hat{n}) = \cos\!\frac{\theta}{2}\, I \;-\; i \sin\!\frac{\theta}{2}\,(\hat{n}\cdot\vec\sigma).
$$

Now rotate all the way around, $\theta = 2\pi$ — a full turn, which classically brings everything back exactly to where it started:

$$
U(2\pi, \hat{n}) = \cos\pi\, I - i\sin\pi\,(\hat{n}\cdot\vec\sigma) = -I.
$$

The spinor does **not** come back to itself. A full $360^\circ$ rotation of space sends $v \mapsto -v$. You have to go around **twice**, a full $720^\circ$, to truly return:

$$
U(4\pi, \hat{n}) = \cos 2\pi\, I - i\sin 2\pi\,(\hat{n}\cdot\vec\sigma) = +I.
$$

That minus sign after one turn is the signature of a spinor, and it is where the leftover sign from the "fixing the phase" section comes home to roost — $U$ and $-U$ really are different, and a $2\pi$ rotation is exactly the operation that swaps them. The half-angle $\theta/2$ in the exponent is the fingerprint of this: physical rotations are being represented in $SU(2)$ at "half speed," so one full physical turn is only a half turn in $SU(2)$, landing on $-I$ rather than $I$.

This is also the concrete reason we insisted on $\mathfrak{su}(2)$ rather than $\mathfrak{so}(3)$ back when we studied real forms. The minus sign cannot be seen by $SO(3)$; the representations that pick it up — the half-integer spins — are representations of $SU(2)$ that do not descend to honest representations of the rotation group. $SU(2)$ is the natural home of spin precisely because it is large enough to carry these two-valued objects that $SO(3)$ cannot.

And this is not a mathematical curiosity with no consequences. The relative sign between a rotated and an unrotated spinor is measurable: in neutron interferometry experiments, a beam of spin-$\tfrac12$ particles is split, one half is rotated by $2\pi$, and when the beams recombine they interfere *destructively* — direct physical evidence that a single turn really did multiply the state by $-1$.

## where this is going

We took three physical facts — probability is conserved, the electron is a two-state system, and a global phase is invisible — and they marched us straight into $SU(2)$ and its algebra $\mathfrak{su}(2)$, the compact real form of the $\mathfrak{sl}(2,\mathbb{C})$ we had already built. The representations turned out to be the spins, and the standard representation turned out to house these strange two-valued spinors.

The one thing we borrowed on credit was the rotation formula $U(\theta,\hat n) = \exp(-i\,\tfrac{\theta}{2}\,\hat n\cdot\vec\sigma)$. Paying that debt — understanding *why* a rotation of $3$-dimensional space acts on spinors this way, why the factor of $\tfrac12$ appears, and how $SU(2)$ sits over $SO(3)$ as a double cover — is the subject of the next chapter. As a bonus, that same relationship will hand us, almost for free, the origin of the cross product from classical mechanics.

S_z v_\text{down} = \frac{\hbar}{2} \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix} \begin{pmatrix} 0 \\ 1 \end{pmatrix} = \frac{\hbar}{2} \begin{pmatrix} 0 \\ -1 \end{pmatrix}.

$$
S_z = \frac{\hbar}{2} \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}.
$$

To represent the measurement of spin of a particle represented by the state vector $v$, we apply the operator $S_z$ to $v$. The result is a new vector that represents the outcome of the measurement:

$$
S_z v = \frac{\hbar}{2} \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix} \begin{pmatrix} \alpha \\ \beta \end{pmatrix} = \frac{\hbar}{2} \begin{pmatrix} \alpha \\ -\beta \end{pmatrix}.
