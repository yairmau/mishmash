We built $\mathfrak{sl}(2,\mathbb{C})$ as the space of traceless $2\times2$ complex matrices, spanned by $E$, $H$, $F$ with *complex* coefficients. That is a generous space — three complex dimensions, which is six real dimensions. Often we do not want all of it. When we talk about rotations, or about spin, we want a smaller, *real* piece: three real dimensions carved out of the six. This chapter is about what that means and how to do it. There are two natural ways to make the cut, and they produce two genuinely different algebras — $\mathfrak{sl}(2,\mathbb{R})$ and $\mathfrak{su}(2)$ — even though both are three-dimensional and both live inside the same $\mathfrak{sl}(2,\mathbb{C})$. The difference between them turns out to be exactly the difference between a boost and a rotation, between a hyperbola and a circle.

## the space we're carving from

Recall that a general element of $\mathfrak{sl}(2,\mathbb{C})$ is

$$
X = a E + b H + c F, \qquad a, b, c \in \mathbb{C}.
$$

Each complex coefficient is really two real numbers, so as a *real* vector space this is six-dimensional. We want to keep a three-dimensional real slice of it. But — and this is the point that is easy to miss — we are not free to keep *any* three-dimensional real slice. We need the slice to still be a Lie algebra, which means it has to be **closed under the bracket**: bracketing two elements of the slice, using only real coefficients, must land us back inside the slice.

Most slices fail this. Take, for instance, the real span of $\{E, F, iH\}$. It is a perfectly good three-real-dimensional subspace, but it is not closed: $[E,F] = H$, and $H$ is not a real combination of $E$, $F$, and $iH$ (it is real and diagonal, while $iH$ is imaginary and diagonal — there is no way to build $H$ from those three with real coefficients). Bracketing threw us out of the slice.

A slice that *is* closed under the bracket, and that regenerates all of $\mathfrak{sl}(2,\mathbb{C})$ once we allow complex coefficients again, is called a **real form**. Concisely: a real form $L$ is a real, bracket-closed subalgebra with

$$
\mathfrak{sl}(2,\mathbb{C}) = L \oplus iL
$$

as real vector spaces. The "$L$" gives you three real dimensions, the "$iL$" gives you the other three, and together they fill out the six. A real form is one self-contained real "half" of the complex algebra.

So the question of this chapter is concrete: which real three-dimensional slices of $\mathfrak{sl}(2,\mathbb{C})$ are bracket-closed? It turns out that, up to isomorphism, there are exactly two. Let's find them.

## the split real form: $\mathfrak{sl}(2,\mathbb{R})$

The most obvious slice is the one you would try first: just demand that the coefficients $a, b, c$ be real. This is the real span of $E$, $H$, $F$:

$$
\mathfrak{sl}(2,\mathbb{R}) = \operatorname{span}_{\mathbb{R}}\{E, H, F\}.
$$

Is it closed under the bracket? Yes, and for a reason that requires no computation: the structure constants of $E, H, F$ are all real,

$$
[H,E] = 2E, \qquad [H,F] = -2F, \qquad [E,F] = H,
$$

so a real combination of these generators brackets to a real combination. Closure is automatic. As matrices, this slice is exactly the traceless $2\times2$ **real** matrices, and the group you get by exponentiating them is $SL(2,\mathbb{R})$: the real, determinant-one, area-preserving transformations of the plane — the very maps we started this whole story with.

This is called the **split** real form. Hold onto the word "split"; the next section will show what is splitting.

## the compact real form: $\mathfrak{su}(2)$

There is a second slice, and it is the one physics cares about. We met it in the [spin chapter](/quantum/su2_and_spinors.html): the traceless **anti-Hermitian** matrices, $\mathfrak{su}(2)$. Let's build it from $E, H, F$ directly and watch what is different.

A basis for this slice is

$$
u_1 = i(E+F), \qquad u_2 = E - F, \qquad u_3 = iH,
$$

or, as matrices,

$$
u_1 = \begin{pmatrix} 0 & i \\ i & 0 \end{pmatrix}, \qquad
u_2 = \begin{pmatrix} 0 & 1 \\ -1 & 0 \end{pmatrix}, \qquad
u_3 = \begin{pmatrix} i & 0 \\ 0 & -i \end{pmatrix}.
$$

Each is traceless and satisfies $u^\dagger = -u$ (anti-Hermitian), so this is $\mathfrak{su}(2)$. And here is the crucial observation: **we needed factors of $i$ to build it.** Two of the three generators, $u_1$ and $u_3$, are imaginary combinations of $E$, $H$, $F$. That is why $\mathfrak{su}(2)$ is a *different* slice from $\mathfrak{sl}(2,\mathbb{R})$, even though both are three real dimensions inside the same complex space. The word "real" in "real form" was never about the matrix entries — those are complex here — it is about only taking real combinations of *these particular* generators.

The slice is bracket-closed. If we use the normalized generators $T_j = -\tfrac{i}{2}\sigma_j$ (which span the same real space), the brackets come out as the rotation structure constants we keep running into,

$$
[T_i, T_j] = \epsilon_{ijk}\, T_k,
$$

with real constants $\epsilon_{ijk}$, so real combinations stay real. Exponentiating gives $SU(2)$, the special unitary group. This is called the **compact** real form — and now let's see why the two forms deserve genuinely different names.

## the same algebra? no.

Both $\mathfrak{sl}(2,\mathbb{R})$ and $\mathfrak{su}(2)$ are three-dimensional real Lie algebras sitting inside $\mathfrak{sl}(2,\mathbb{C})$. It is tempting to think they must therefore be "the same algebra in disguise." They are not. There is no way to match one to the other by a change of real basis — they are genuinely distinct — and the cleanest way to see it is to ask what happens when you exponentiate a generator.

For a traceless $2\times2$ matrix, the Cayley–Hamilton theorem gives a small miracle: $X^2 = -\det(X)\, I$. So the single number $\det(X)$ controls everything. If $\det(X) > 0$, then $X^2 = -(\text{positive}) I$ behaves like $i$, and exponentiating gives **circles** — bounded, periodic motion that returns to where it started. If $\det(X) < 0$, then $X^2 = +(\text{positive}) I$, and exponentiating gives **hyperbolas** — unbounded motion that runs off to infinity.

Watch this play out. In $\mathfrak{sl}(2,\mathbb{R})$, the generator $E - F$ has $\det = 1 > 0$, and indeed

$$
e^{\theta(E-F)} = \cos\theta\, I + \sin\theta\,(E-F)
$$

is an ordinary rotation, periodic with period $2\pi$. But the generator $H$ has $\det = -1 < 0$, and

$$
e^{tH} = \cosh t\, I + \sinh t\, H = \begin{pmatrix} e^{t} & 0 \\ 0 & e^{-t} \end{pmatrix}
$$

stretches one axis and shrinks the other, running off to infinity as $t$ grows. So $\mathfrak{sl}(2,\mathbb{R})$ contains **one** rotation direction and **two** stretching (boost) directions. That is the "split": the group has both bounded and unbounded parts.

Now look at $\mathfrak{su}(2)$. Every one of its generators has positive determinant — $\det(u_1) = \det(u_2) = \det(u_3) = 1$ — so *every* direction exponentiates to a circle. $SU(2)$ is all rotations, no boosts; nothing runs off to infinity. This is what **compact** means: the group is a bounded, closed set (topologically $SU(2)$ is a $3$-sphere).

The rigorous invariant behind this intuition is the **trace form** $\langle X, Y\rangle = \operatorname{tr}(XY)$ (proportional to the Killing form). Square each basis generator and read off the sign of the trace. The comparison is striking, because the two forms are built from almost the same ingredients:

$$
\begin{array}{lll}
\mathfrak{sl}(2,\mathbb{R}): & \{\,H,\ E+F,\ E-F\,\} & \operatorname{tr}(X^2) = +2,\ +2,\ -2 \\[4pt]
\mathfrak{su}(2): & \{\,iH,\ i(E+F),\ E-F\,\} & \operatorname{tr}(X^2) = -2,\ -2,\ -2
\end{array}
$$

The two lists differ in exactly one respect: $\mathfrak{su}(2)$ takes the two positive-trace directions of $\mathfrak{sl}(2,\mathbb{R})$ — namely $H$ and $E+F$ — and multiplies them by $i$. Since $\operatorname{tr}\big((iX)^2\big) = -\operatorname{tr}(X^2)$, that factor of $i$ flips each $+2$ into a $-2$. The one direction both forms share, $E - F$, was already negative — it is the rotation, the compact direction that needs no fixing.

So the trace form of $\mathfrak{sl}(2,\mathbb{R})$ is **indefinite**, with signature $(2,1)$ — two plus signs, one minus — while the trace form of $\mathfrak{su}(2)$ is **negative definite**, signature $(0,3)$. A change of real basis can never turn an indefinite form into a definite one (the number of plus and minus signs is an invariant), so no isomorphism between the two real forms can exist. They are different algebras, full stop.

And the physics is sitting right there in the sign counts. A negative-definite form with three compact directions is $\mathfrak{so}(3)$: rotations of ordinary three-dimensional space, three of them, in the $xy$, $yz$, $zx$ planes. An indefinite $(2,1)$ form with one compact and two non-compact directions is $\mathfrak{so}(2,1)$: the symmetries of a space with one time and two space dimensions — one rotation (in the spatial plane) and two boosts (mixing time with each space axis). So the two real forms are

$$
\mathfrak{su}(2) \;\cong\; \mathfrak{so}(3) \qquad\text{and}\qquad \mathfrak{sl}(2,\mathbb{R}) \;\cong\; \mathfrak{so}(2,1),
$$

and the whole compact-versus-split distinction is the distinction between the geometry of a sphere and the geometry of spacetime.

## what picks out each slice

There is a tidy way to characterize each real form intrinsically, rather than by writing down a basis — and it explains why there are exactly two. Each real form is the set of matrices left *fixed* by a certain operation on $\mathfrak{sl}(2,\mathbb{C})$.

Consider two operations. The first is ordinary complex conjugation of the entries, $X \mapsto \bar{X}$. Its fixed points — the matrices equal to their own conjugate — are the ones with real entries, which is exactly $\mathfrak{sl}(2,\mathbb{R})$. The second is the negative conjugate-transpose, $X \mapsto -X^\dagger$. Its fixed points are the matrices with $-X^\dagger = X$, the anti-Hermitian ones, which is exactly $\mathfrak{su}(2)$.

$$
\mathfrak{sl}(2,\mathbb{R}) = \{\, X : \bar{X} = X \,\}, \qquad
\mathfrak{su}(2) = \{\, X : -X^\dagger = X \,\}.
$$

Both operations are *antilinear* (they conjugate the $i$'s) and both square to the identity — they are involutions. Every real form of a complex Lie algebra arises as the fixed points of such an involution, and for $\mathfrak{sl}(2,\mathbb{C})$ these two are, up to isomorphism, all there are. So the two forms are not an arbitrary pair we happened to find; they are the complete list, each cut out by a natural involution.

## complexification: the way back

The defining property of a real form is that you can climb back up from it. Allow complex coefficients again — this is called **complexifying** — and either form regenerates all of $\mathfrak{sl}(2,\mathbb{C})$:

$$
\mathfrak{sl}(2,\mathbb{C}) = \mathfrak{sl}(2,\mathbb{R}) \oplus i\,\mathfrak{sl}(2,\mathbb{R}) = \mathfrak{su}(2) \oplus i\,\mathfrak{su}(2).
$$

Both real forms are skeletons of the same complex body; they are just two different ways of choosing which half is "real."

The second splitting has a beautiful physical reading. The elements of $\mathfrak{su}(2)$ are anti-Hermitian, so the elements of $i\,\mathfrak{su}(2)$ are Hermitian. Thus

$$
\mathfrak{sl}(2,\mathbb{C}) = \underbrace{\mathfrak{su}(2)}_{\text{anti-Hermitian}} \oplus \underbrace{i\,\mathfrak{su}(2)}_{\text{Hermitian}} = \underbrace{(\text{rotations})}_{\text{compact}} \oplus \underbrace{(\text{boosts})}_{\text{non-compact}}.
$$

Rotations plus boosts is the Lorentz algebra: as a real six-dimensional Lie algebra, $\mathfrak{sl}(2,\mathbb{C})$ is $\mathfrak{so}(1,3)$, the symmetries of four-dimensional spacetime. Taking the compact real form $\mathfrak{su}(2)$ is, in this light, the act of keeping only the rotations and discarding the boosts. That is exactly what we did, without saying so, when we motivated $\mathfrak{su}(2)$ from probability conservation in the spin chapter: unitarity kept the compact, bounded directions and threw the boosts away.

## where this goes

We started with one complex algebra and found it has two real skeletons. One of them, $\mathfrak{su}(2) \cong \mathfrak{so}(3)$, is the compact form, the home of rotations and spin; we have already used it to build spinors, and in the [next chapter](/quantum/su2_so3_double_cover.html) we will pin down precisely how $SU(2)$ sits over the rotation group $SO(3)$ as a double cover — and get the cross product of classical mechanics as a parting gift. The other, $\mathfrak{sl}(2,\mathbb{R}) \cong \mathfrak{so}(2,1)$, is the split form, the home of boosts and hyperbolic geometry, and it has its own long story: it is the symmetry algebra of the hyperbolic plane and of $2+1$-dimensional relativity. Same three letters, $E$, $H$, $F$ — but which combinations you are allowed to take, and which of them you multiply by $i$, decides whether you are doing geometry on a sphere or on a hyperboloid.
