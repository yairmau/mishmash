## conditional independence theorem

I have three random variables: $X$, $Y$ and $Z$. We assume their joint probability density to by a gaussian with zero mean in all directions, and a (symmetric) precision matrix called $\Lambda$. We can write the pdf as

$$
p(X,Y,Z) \propto \exp\left( -\frac{1}{2} x^T \Lambda x \right),
$$

where $x=(X,Y,Z)^T$ is a column vector.

::: {#thm-triangle}
The matrix component $\Lambda_{XZ}=0$ iff $X\perp Z \mid Y$.
:::

::: {.proof}
Let's write out $x^T \Lambda x$ explicitly. Our convention is that $\Lambda_{ij}$ denotes the matrix component in the row $i$ and column $j$.

\begin{align*}
x^T \Lambda x =&
\begin{pmatrix}X&Y&Z\end{pmatrix}
\begin{pmatrix}
  \Lambda_{XX}&\Lambda_{XY}&\Lambda_{XZ}\\
  \Lambda_{YX}&\Lambda_{YY}&\Lambda_{YZ}\\
  \Lambda_{ZX}&\Lambda_{ZY}&\Lambda_{ZZ}
\end{pmatrix}
\begin{pmatrix}X\\Y\\Z\end{pmatrix} \\
=&
\begin{pmatrix}X&Y&Z\end{pmatrix}
\begin{pmatrix}
\Lambda_{XX}X+\Lambda_{XY}Y+\Lambda_{XZ}Z \\
\Lambda_{YX}X+\Lambda_{YY}Y+\Lambda_{YZ}Z \\
\Lambda_{ZX}X+\Lambda_{ZY}Y+\Lambda_{ZZ}Z
\end{pmatrix} \\
=&
X(\Lambda_{XX}X+\Lambda_{XY}Y+\Lambda_{XZ}Z) + \\
&Y(\Lambda_{YX}X+\Lambda_{YY}Y+\Lambda_{YZ}Z) + \\
&Z(\Lambda_{ZX}X+\Lambda_{ZY}Y+\Lambda_{ZZ}Z) \\
=& X^2 \Lambda_{XX} + Y^2 \Lambda_{YY} + Z^2 \Lambda_{ZZ} + \\
& 2XY \Lambda_{XY} + 2XZ \Lambda_{XZ} + 2YZ \Lambda_{YZ},
\end{align*}

where we used in the last step the fact that $\Lambda$ is symmetric, therefore $\Lambda_{ij}=\Lambda_{ji}$.

The next step is to condition on $Y$. This means that we treat the random variable $Y$ as a known number, we'll call it $y$. We will now substitute the result of the last derivation, conditioned on $Y$, into the expression for the joint pdf.

\begin{align*}
p(X,Y,Z) \propto & \exp\left( -\frac{1}{2} x^T \Lambda x \right) \\
=& \underbrace{\exp\left[ -\frac{1}{2} \left( X^2 \Lambda_{XX} + 2Xy \Lambda_{XY} \right) \right]}_{\text{depends only on }X} \times \\
& \underbrace{\exp\left[ -\frac{1}{2} \left( Z^2 \Lambda_{ZZ} + 2yZ \Lambda_{YZ}\right) \right]}_{\text{depends only on }Z} \times \\
& \underbrace{\exp\left[ -\frac{1}{2} 2XZ \Lambda_{XZ} \right]}_{\text{depends only both }X,Z} \times \\
& \underbrace{\exp\left[ -\frac{1}{2} y^2 \Lambda_{YY} \right]}_{\text{constant}} \times \\
\end{align*}

Two random variables $A$ and $B$ are said to be *indepentent* if

$$
p(A,B) = p(A)\cdot p(B).
$$

They would be said to be *conditionally independent* (on $C$, say) if

$$
p(A,B \mid C) = p(A \mid C)\cdot p(B \mid C).
$$

:::
